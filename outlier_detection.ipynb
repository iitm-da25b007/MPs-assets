```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Configuration
DATA_FOLDER = Path("clean")

FILES = [
    "andhra_2014.csv",
    "andhra_2019.csv",
    "andhra_2024.csv",
    "bihar_2014.csv",
    "bihar_2019.csv",
    "bihar_2024.csv",
    "gujarat_2014.csv",
    "gujarat_2019.csv",
    "gujarat_2024.csv"
]

ASSET_COLUMN = "Total Assets"
CANDIDATE_COLUMN = "Candidate"

IQR_MULTIPLIER = 1.5


# IQR detection function
def detect_outliers_iqr(data, multiplier=1.5):

    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)

    IQR = Q3 - Q1

    lower_bound = Q1 - multiplier * IQR
    upper_bound = Q3 + multiplier * IQR

    outlier_mask = (
        (data < lower_bound) |
        (data > upper_bound)
    )

    return outlier_mask, lower_bound, upper_bound


# Process each file
def process_file(file_path):

    df = pd.read_csv(file_path)

    # Convert assets to numeric
    asset_values = pd.to_numeric(
        df[ASSET_COLUMN].astype(str)
        .str.replace(r"[₹,]", "", regex=True)
        .str.strip(),
        errors="coerce"
    )

    valid_mask = (
        asset_values.notna() &
        np.isfinite(asset_values)
    )

    valid_df = df.loc[valid_mask].copy()
    values = asset_values.loc[valid_mask].to_numpy()

    # Detect outliers
    outlier_mask, lower, upper = detect_outliers_iqr(
        values, IQR_MULTIPLIER
    )

    valid_df["IQR_Outlier"] = outlier_mask

    # Plot
    plt.figure(figsize=(12, 6))

    normal_mask = ~outlier_mask

    plt.scatter(
        np.where(normal_mask)[0],
        values[normal_mask],
        color="royalblue",
        label="Normal",
        alpha=0.65,
        s=30
    )

    plt.scatter(
        np.where(outlier_mask)[0],
        values[outlier_mask],
        color="red",
        label="Outlier",
        alpha=0.85,
        s=45
    )

    # Show IQR boundaries
    plt.axhline(
        lower,
        color="green",
        linestyle="--",
        label="Lower Bound"
    )

    plt.axhline(
        upper,
        color="orange",
        linestyle="--",
        label="Upper Bound"
    )

    # Log-like scaling for better visibility
    plt.yscale("symlog", linthresh=1e5)

    plt.title(f"IQR Outlier Detection - {file_path.stem}")
    plt.xlabel("Candidate Index")
    plt.ylabel("Declared Assets (₹)")

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Print results
    outliers = valid_df.loc[outlier_mask]

    print("\n", "=" * 60)
    print(f"FILE: {file_path.name}")
    print("=" * 60)

    print(f"Total candidates: {len(valid_df)}")
    print(f"Outliers detected: {len(outliers)}")
    print(f"Lower bound: {lower:,.2f}")
    print(f"Upper bound: {upper:,.2f}")

    if len(outliers) > 0:
        print("\nDetected Outliers:")

        print(
            outliers[
                [CANDIDATE_COLUMN, ASSET_COLUMN]
            ].to_string(index=False)
        )

    return valid_df


# Run for all nine files
all_iqr_results = {}

for filename in FILES:

    file_path = DATA_FOLDER / filename

    if not file_path.exists():
        print(f"File not found: {filename}")
        continue

    all_iqr_results[filename] = process_file(file_path)
```
