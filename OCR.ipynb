```python
import pandas as pd
```

```python
import easyocr

reader = easyocr.Reader(['en'])

def ocr_assets(image_url):
  if not image_url.startswith("http"):
    return image_url

  results = reader.readtext(image_url)
  if not results:
      print("failed to fetch"+f"{image_url}")
      return image_url
  else:
      text_fragments = [item[1] for item in results]

      final_text = " ".join(text_fragments)
      return final_text
```

```python
from google.colab import files
def final_downloader(filename):
  df = pd.read_csv(filename)

  assets = df['Total Assets'].apply(ocr_assets)
  liabilities = df['Liabilities'].apply(ocr_assets)

  df['Total Assets'] = assets
  df['Liabilities'] = liabilities

  df.to_csv(filename, index=False)
  files.download(filename)
```

```python
final_downloader("andhra_2014.csv")
```
